```python
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
quantization_config = BitsAndBytesConfig(
    load_in_8bit=True  # Alternatively, use this line for 8-bit quantization
)
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Meta-Llama-3.1-8B", use_auth_token=True)
model = AutoModelForCausalLM.from_pretrained("meta-llama/Meta-Llama-3.1-8B",use_auth_token=True,device_map="auto", quantization_config=quantization_config)
```

```python
def generate_output(prompt,max_tokens,do_sample_set,temperature_score,penalty,return_number,skip_special_token  ):
    print(f"Input Text: {prompt}, Max Tokens: {max_tokens}, Do Sample: {do_sample_set}, Temperature: {temperature_score}, Penalty: {penalty}, Return Number: {return_number}, Skip Special Tokens: {skip_special_token}")
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    generated_outputs = model.generate(
    **inputs,
    max_new_tokens=max_tokens,         # Generate up to 1048 new tokens
    do_sample=do_sample_set,              # Use sampling for diversity
    temperature=temperature_score,             # Moderate randomness
    #top_p=0.9,                   # Nucleus sampling
   # top_k=50,                    # Consider top 50 tokens at each step
    repetition_penalty=penalty,      # Penalize repetition
    eos_token_id=tokenizer.eos_token_id,  # Stop generation at EOS token
    num_return_sequences=return_number      # Return one sequence
)
    # Decode the generated outputs
    generated_text = tokenizer.decode(generated_outputs[0], skip_special_tokens=skip_special_token)

    # Print the generated text
    print("\n output Generated\n")
    return generated_text
#text="ہم نے ابھی۔ویڈیو ہے۔لاش فائل ڈاؤن لوڈ ہو رہی میرا رکن ابھی بھی ٹائم لگ جائے گا۔لیکن وہ یاد وہ آج دونوں نے اپ لوڈ کرنے کا عزم کر لیجیے گا کیونکہ وہ ہماری میٹنگ کے نوٹس بن گیا ہے اس کے ڈیٹ آئی تھنک ویک اینڈ شئیر ود دیم اینڈ دا وے رائٹ انس۔میٹنگ۔نہیں۔نہیں۔کیونکہ اگر آپ نے مینشن کی ہوئی ہیں کچھ خود سے سمری جنریشن میں انہوں نے مجھے اگر کبھی بہت ساری جو میٹنگز نے رینیوئل لینگویج والی ہیں سمری اسی وقت ہی جنریٹ کر پائے ٹرانسکرپٹ گروی رکھ گئے بڑ کرتا ہے لیکن وہ زیادہ چاہیے وٹ ڈو دے ہیو دیئر بی ریئلی ڈونٹ نو۔نام شامل ہے۔از ناٹ ان یور کیریئر سمجھتے ہیں جیسے رحمن لکھا ہوا تھا نا وہاں پہ تو اس کا نام کا جو تھانہ ٹیمز اس پہ آ رہا تھا۔اس کا نام آ رہے ہیں لیکن وہ ٹرانسفر کے اندر اس۔ٹائم۔خود ایک چھوٹی چیز ہے لیکن ویسے اوور آل کانٹیکسٹ بنایا ہے اس کو ہم فائن ٹیوننگ کے وہ انگلش تھی نا کوئی چیک اینڈ بیلنس جو ہے ناں لینگویج اردو پہ دے دوں گا تو پراببلی مجھے اتنا اچھا چلیں تو آئی نو علی وہ سٹارٹنگ پوائنٹ ہے جس نے کہا ہے کہ آ گئے ہیں۔:م:۔اس نے گل کا ایک پوائنٹ اور اس کے کہ ہم درمیان میں ہم اپنے ماڈل کی بجائے مائع جیسی اس سے زیادہ بہتر ہو جائے۔باقی کام ہم نے کیا۔کیونکہ ابھی جو بھی کوئی آپ کو آپ چیلنج کرے گا۔وہ چیلنجز کسی سے پیار کرے گا۔نہیں لینگویج ہے وہ ماڈل کے مطلب ابھی ادھر بھی موت ہی وہ ماڈل کی ایکوریسی کو ہم ایسے :م: ابھی صرف ان کمپیریزن سے جائز نہیں کر پا رہا ہے کہ کس کا ماڈل زیادہ اچھا کام کر رہا ہے پرفارم کر رہا ہے۔میں یونیورسل وہ کام وہیں پہ ملاقات ہوئی میرے اور رخ کرنے لگے۔نہیں ہمیں ایشو نا کریں میں یہ کہہ رہا ہوں کہ ہم نے آبجیکٹیو یہ دیکھنا ہے کہ جو دوسرے ماڈل نے سمری کی سمجھ میں نہیں پڑھ کے آ رہی ہے اور ہم نہیں سمجھتے کہ اردو پڑھ کیا کیا میٹنگ میں یہی بات ہو رہی تھی کہ نہیں ہونی چاہیے یہاں دس ایز ون پوائنٹ۔لیکن ہم نے ایک جنریٹ کرنی ہے کہ فائر پلایا جب اردو سمری جنریٹ کرتا ہے وہ فیلیئر ہیں ٹھیک ہے لیکن اگر اسی میٹنگ کی وہ انگریزی سمری جنریٹ کرنا ہے کیا وہ ہم سے ایکٹ پا رہا ہے انفیریئر ہے اس پیریڈ ہے۔نیکسٹ تھنگ ایز کہ وہ ایکشن آئٹمز جو نکال رہا ہے۔ان۔نہیں۔ہاں۔:م: اس پہ :م: یہ میٹنگ اینڈ کریں یہیں پہ آ جائے گا۔نہیں۔اس نے۔اس کی گئی ہے۔ڈسکشن۔ٹھیک ہے تو وہ کر سکتے ہیں تو یہاں نیوزی لینڈ کر رہا ہے۔جی نہیں۔یہ وہ۔جی نہیں۔ان کا کہنا ہے اس کو لے کر ایک لنک میوزک نوٹس کریں۔اس قوم نے اپنے اس پہ کیا ہوا ہے سسٹم پہ :م:۔اچھا یہ نیچے اس میں دیکھیں اور نیچے ہائی لائٹ کیا نکالتا ہے۔یہاں لنکس ہیں ٹھیک ہے۔اور نیچے ایکشن آئٹمز ہوں گے۔لیٹس کم ٹو ڈے افیئرز اینڈ آر میں اسی میں۔اور پہلے سے ہی نہیں اور لوگوں نے :م:۔پائلٹ نے کچھ اور۔نہیں نہیں۔کہ جو پوائنٹس۔جی یہ۔جی نہیں۔نہیں۔اس کے۔جی نہیں۔چلیں جی ایک دو چیزیں تو نکالیں ناں آپ نے ویسے :م: یہ آپ نے دیکھ لیا ہے اسی اب آپ باہر فیض کو اردو میں کنورٹ واپس لبنان میں لینڈنگ کی تھی اپنی فیکٹ ہے اس کی اب دوبارہ کر دیں نا دیں کہ اردو میں کیا کہتا ہے آپ کو اسی کو کسی میٹنگ میں کیسے بتاتا ہے۔فیریئر ہو گئی پٹیشنز پڑی۔لیکن جس میں وہ کچھ میک سینس کرتا ہے بنا لیتا ہو گا۔جو کہ میرا نہیں خیال ہے ہو گا۔میں نہیں آتے کہ وہ کرکٹر آف دی انڈسٹریل پارک نا ہو جسٹ گوئنگ ٹو گو ان دی یوز کیسز۔لائک :م: اردو کے ٹیکس بچائے گا وہ آڈیو آئے گی اور وہ پھر بھی میک سینس کرتا ہے ٹھیک ہے :م:۔نہیں۔موصوف نے حوصلہ نہیں ہو گا اردو میں بھی اگر کیری کریں سمندر میں ماہی یعنی جس ٹیبل تو دور بیٹھے ٹرانسلیشن آف دا انگلش ٹو اردو آف فیصلہ ایس ایس آر کا میرٹ ہے۔موسم بہت ضروری ہے ہاں۔تیرے سو گئی اہل سمجھ نہیں آئے گی۔نہیں۔وہ۔اسی کا ہے۔نہیں۔نہیں۔نہیں نہیں نہیں وہ تو میرے گھر آنے دیں کہ وہ کسی بھی فیکٹری لگی ہوئی ہے کوئی بات ہے ابھی بھی پی لو بندہ ہے سیل میں بند ہے وہ اپنی ٹیم کی اصلاح کرے دینے کی جو ہے نا وہ ہاف ڈے 20 منٹ میں آدھے گھنٹے میں گاڑی کی۔چلاتے رہے اور پارٹی فار دا سن سکتا ہے ٹھیک ہے پانچ پانچ منٹ کی مرسیڈیز آ رہی ہے سب کی انعام جس نے نہیں کی جاتی ہے۔نہیں۔نہیں۔اچھا اچھا۔کیوں نہیں ہے۔سے آ رہی ہے۔نہیں۔یہاں پہ۔نہیں۔س کیا۔یہ آپ نے اپ لوڈ کی گئی ہے کہ آپ نے میری کیا میری ٹیم ٹیم میٹس ہے اور اردو پہ آئیڈیا۔اردو ہے۔اور ایک نیز اردو میں انسان۔نہیں۔بیک اپس ہو گئے۔نہیں۔انہوں نے کہا ہے کہ ہے۔چلیں اس کو کنسیڈر کیا جا سکتا ہے۔کسی میئر کا پیکج ہو سکتا ہے۔:م:۔گی۔سعودی وانا پروسیجر کیا ہو گی اس میں فیڈ بیک کو سٹور کر رہے ہیں نا کہ جی۔ادھر زیادہ جو :م:۔اب وہ دو تین چیزیں جو ہمیں ایڈ کرنے والی ہیں وہ تو وہ ہے کہ کوئی ٹیمز اور ان کی ہونی چاہیے۔وہ بھی آپ نے۔ٹھیک ہے یا نہیں۔السلام علیکم۔میری آواز۔اچھا۔صحیح گڈ ہو گی۔دس از فار آل۔وچ ایز بیڈ۔یہاں بیٹھ کر اپنے اینڈ پہ کلیئر نہیں ہوئے۔:م:۔ان کے ساتھ۔اس طرح۔اس میں۔میں۔اس کی بیوی آپس میں کمپیٹیشن ہے تو کوشش کرتے ہیں۔ہیلپ آؤٹ کر دیں اس میں جھوٹی قسم سے انٹیگریٹ کرنا ہے۔نہیں۔نہیں۔شور سر۔بالکل بیکوز دے آر ٹی اے پی آئی ٹرائیڈ ٹو ڈو اٹ ایز ایون ہیئر آپ نے دیکھا ہے آڈٹ ہوا ہے۔آپ دیکھیں کہ آپ نے ادھر سے کہا۔آپ نے ایک رپورٹر۔سو یو ہیو ٹو فگر آؤٹ کہ یو نو فریش :م: فائر نا اہل کیا جاتا ہے یہ ایکٹ اگر یہ رپورٹ :م:۔وہاں پہ سکون ہوتا۔نئے۔جب میں ان کی سٹینڈرڈ ہے فالٹ لائن سے کینیڈا کے صوبے بھی اس پہ :م: وہ گارنٹی ایک یوزر ہے۔پیریڈ فائر گلاس فریش فائر ٹائمز یوز ہو جاتا ہے۔جس کا حوالہ کر دیتے ہیں۔اور وہ ریکارڈ فراہم کر کے بیٹھ کر وہ آڈیو ریکارڈ کرتا ہے اور پھر اپنی بیٹنگ بہتر آل آل ڈن میاں چنوں۔جی سر وہ تو۔ویسے ہی حسن ہی ہوئے بولا۔سے چل رہا ہے۔کیا ہوا ہے وہ اٹھانے کی ایسی تشریح۔جوائننگ اس۔انہوں نے یہ کیوں کیا۔کیا وہاں سے۔ساری ڈیٹیلز ہیں۔کچھ میٹنگز لکھنے لگی تھی سب کو مل سکتی ہے۔وہ گلاس خیر ابھی چل رہا ہے یہ پہلا کام یہ کرنا ہے کہ بیسکلی میں یہ کہوں کہ میری میں آپ کے خیال میں آئی اور میں اپنی سائیڈ پہ ریکارڈنگ آن کر۔میرے پاس آڈیو آئے گی پھر میں نے اسے وارننگ دے دی ہے وہ ہمارے سسٹم کے اندر چلا جائے۔بھی کیا ہے ٹھیک ہے بہت زیادہ دیر واٹس ایپ آڈیو ریکارڈنگ کرتا ہے ریکارڈنگ جو ہے وہ آپ کے پیس یا سانگ کی ریکارڈنگ نہیں ہو رہی ہوتی اس کا اپنا ریکارڈ اچھا رہا ہوتا۔وہ ویٹ نہیں کرتا ایٹ ایز اپنی ریکارڈنگ کر کے مجھے دے اور پھر میں اس قابل نہیں۔ٹھیک ہے آپ خود ہی ریکارڈنگ کر رہا ہے وٹس دا وے۔کر رہا ہوتا ہے جیسے ادا کیا ہے وہ ریکارڈنگ اپنے پاس سیو کرتا ہے اور ان کے بازی کی گئی جو ایریا ہے یہاں پہ آپ کو آڈیو میں آڈیو اپ لوڈ کریں ادھر آپ آڈیئنس کو گالی دے دیں گے اور ثابت ہونے کا نام ہے جو میٹنگ لیڈ کریں گے اور وہ اگر آڈیو ہے پروسس کر کے۔پندرہ دن بعد بتا دیا کہ ناسٹیلجیا دفاعی پیدا نہیں کرتا ہے پہلے پانچ دس منٹ بعد ہاتھ مل مل کر دی گئی ویلج کونسل ہو گی اور اسمبلی شبلی یہ تو یہ چیز ہونی چاہئیے نا کہ کور کرے گا کون ڈاؤن لوڈ کرے گا اور یہ کریے گا :م: آئی مین دیس ایز آل آئی ایم ویری سیریئس۔اگر میرا کام آسان ہے۔دس از ویری گڈ بیکوز وی نیڈ ٹو بی ڈن اکارڈنگ آڈیو اسمبلی لیے۔دیٹس دیٹس سمتھنگ جائے اور۔مطلب وہ آڈیو لیں۔ٹھیک ہے۔ہے وہ ہے آپ اس کی ایکوریسی کا کمپیریزن کرنا پڑتا وہ اور ان وائس کریں۔این آئی نئی دہلی۔ایون مور ہیوی سینز ڈیٹ واز اے لاٹ ڈیٹ وی ول فور اس :م: اس میں پیچھے نہیں۔اوکے میم۔ہے۔نہیں۔نہیں۔جیسے۔جی اس کی کائنڈلی۔تو اس میں۔وہ۔نہیں بس یہ ہی ہے تو۔نہیں شکریہ۔جس میں وہ ہے۔:م: موومنٹ نہیں ہے۔کیا۔کہ وہ اسی نئی حکومت بننے نہیں سیگمنٹس جو کہ۔وہ ایک ہی ہے۔جب دوبارہ وہ سیٹ ہو جائے۔ایف آئی آر بی آئی میں زیادہ ہے۔آئین اس کی سیگمینٹیشن نہیں آئی۔اوور لیپنگ سیگمنٹس کو کیا ادھر ایف آئی آر ایف کے چنگل سے میں بریک ٹائم تو اس میں ٹائم :م: لنگوسٹ کی طرف دیکھیں تو اس طرح نیلی زیادہ :م: جو ہمارے پہ نہیں آئے۔نہیں بس۔ان کی پارٹی میں جو اردو اور انگلش دونوں کہ میں کہیں گے تو نکال لیں۔:م: انیشیئلی کاسٹ یہ آپ کو ایک تو میں نے کہا یار لوگ اس کو سپیسفک بھی کیا ہے۔رشی کا ہے۔یس میم یہ پڑھنے والا ہے۔وہ۔جس طرح ہم نے تو کہ اس نے کافی پیش آیا۔اور یہ بھی کہہ دیا میں نے اس طرح کی وہ میٹنگ ہے ناں وہ :م: وہ وہ وہ لیکن آپ کے بزنس کیس میں استغاثہ کے تین الگ الگ گھروندے بنتے جا رہے چوتھے منظر صرف انگریزی بولتا رہا ہے اور ہم نے نہیں جانا ہے تو اس کیس کو ہینڈل کرنا پڑے۔پڑھنا بزنس کے ساتھ کھڑی کرو۔:م:۔:م: ہر ڈومین سپیسفکلی اس کی سزائیں ایسی قسم لی کہ میں جیل خانہ جات نے میڈیکل وین کے لیے انوالو کریں گے ہے بہت فائدے مند دیئر دیئر از ناٹ ایٹ آل میں آئی تھینک یو۔بیجنگ۔اینڈ ابھی جو موسم میں بورڈ میں :م: دس :م: کتنی ہے کمر سر میم نہیں اور وہ جس کے بعد آپ بیچیں گے کہ ملٹری وہ زیادہ بہتر ہے کہ آپ۔وہ ٹیم جو ہے یہ کبھی نہیں سنا دیتے ہیں۔ٹھیک ہے میں :م: جو وائس پریزیڈینٹ ہے۔ٹھیک ہے اور اس نے تو بس وہ ویسے یوز کیا ہے ڈالی جائے تو بائی دا وے سوا گھنٹے اچ ڈال دیا۔ٹھیک ہے نا ہم اس کی سمریز جنریٹ کریں گے کہ جی اس کے اندر یہ کہا گیا ہے نا ایک ماڈیول ہے نا ایک ایسی کوئی :م: جو میٹنگ ہے وہ :م:۔وی کین پلے اینڈ فیئر کرائیں اور خاص طور پہ اگر آپ کے۔پارٹی ڈال رہی ہے۔وہ لیلیٰ :م: جو ہے وہ ٹیک اول میئرز۔:م: جی نہیں تھینک یو۔اپنی ہے :م: سیمی فائنل میں کوریا کے تجزیہ اور میں لڑائی میں قائم کی گئی اس قسم کا اگر ان کے کتنے ہی ایک بندہ ہے جس کا سارا کام ہے کہ اس نے نامور تجزیہ نے اس کی زندگی آسان ہو جائے پھر فنانشل ماڈل ہے۔ظاہر ہوتا ہے :م: یہ جو ہمارے اس ٹائم نا یہ جو فنانشل اینالسٹ ہوتے ہیں نا ان یہ بہت استعمال کرتے ہیں پہلے انہوں نے گلوبل ٹائمز یہ "
#Summary_Genration(text)
```
